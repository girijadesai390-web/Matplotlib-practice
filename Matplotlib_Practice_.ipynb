```python
from matplotlib.lines import lineStyles
#Histogram
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
Marks=np.random.normal(loc=35,scale=8,size=1000)
plt.figure(figsize=(8,5))
plt.hist(Marks,bins=10,color='#0ea5e9',edgecolor='white',alpha=0.8)
plt.title("Distribution of Marks",fontsize=14,fontweight='bold',pad=15)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid(linestyle='--',alpha=0.2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed (42)
class_A = np.random.normal(90, 10, 100)
class_B = np.random.normal(70, 7, 100)
class_C = np.random.normal(50, 15, 100)
data_to_plot = [class_A, class_B, class_C]
plt.figure(figsize=(8, 5))
plt.boxplot(
data_to_plot,
labels=['Class A', 'Class B', 'Class C'],
patch_artist=True
)
plt.title('Marks Distribution Across Classes', fontsize=14, fontweight='bold', pad=15)
plt.ylabel('Marks', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
